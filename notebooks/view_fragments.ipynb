```python
from rdkit import Chem
from rdkit.Chem import Draw

from bcosgnn.fragmentation import find_brics_fragments
aspirin = "CC(=O)Oc1ccccc1C(=O)O"

# mol = Chem.MolFromSmiles(random.choice(smiles_train))
mol = Chem.MolFromSmiles(aspirin)
frags, bonds = find_brics_fragments(mol)
```

```python
Draw.MolToImage(mol, size=(600, 600))
```

```python
from rdkit.Chem import Draw

Draw.MolsToGridImage(
    [fragment.mol for fragment in frags],
    molsPerRow=4,
    subImgSize=(300, 300),
)
```
